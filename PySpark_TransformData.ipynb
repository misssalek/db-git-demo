```python
df2=spark.table('gizmobox.bronze.py_customers');
display(df2);
# remove null customer id s
df2=df2.filter(df2.customer_id.isNotNull())
display(df2);
# remove null email s


```

```python
df_filtered = df2.where(df2.customer_id.isNotNull())
display (df_filtered)
```

```python
df_distinct = df_filtered.distinct()
# df_distinct = df_filtered.dropDuplicates()

display(df_distinct)
```

```python
from pyspark.sql import functions as F

df_max_ts = df_distinct.groupBy('customer_id') \
                       .agg(F.max('created_timestamp').alias('max_created_timestamp'))
```
